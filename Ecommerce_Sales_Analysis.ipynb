```python
# import Libraries
import pandas as pd
import random
```

```python
# synthetic ecommerce dataset.

random.seed(42)

# Define data parameters
regions = ['North', 'South', 'East', 'West']
categories = ['Electronics', 'Clothing', 'Home & Garden', 'Sports', 'Books']
salespersons = ['Alice', 'Bob', 'Carol', 'David', 'Emma', 'Frank']

# Generate 200 sales transactions
data = {
    'transaction_id': range(1001, 1201),
    'region': [random.choice(regions) for _ in range(200)],
    'category': [random.choice(categories) for _ in range(200)],
    'salesperson': [random.choice(salespersons) for _ in range(200)],
    'sales_amount': [round(random.uniform(50, 5000), 2) for _ in range(200)],
    'customer_id': [random.randint(5000, 5100) for _ in range(200)]
}

df = pd.DataFrame(data)

print(df.head(10))
print("\nDataset shape:", df.shape)
```

# Task 1: Basic Grouping and Single Aggregations

## 1.1 Total sales per region




```python
# Identify which region generates the highest revenue.

region_sales = df.groupby('region')['sales_amount'].sum().reset_index()
(region_sales)
```

## 1.2 Count Transactions Per Category

```python
# Identify which product category sells most frequently
category_transactions = df.groupby('category')['transaction_id'].count().reset_index()

category_transactions
```

## 1.3 Average Sales per Salesperson

```python
# To analyze average sales value generated by each salesperson

avg_salesperson = df.groupby('salesperson')['sales_amount'].mean().reset_index()
avg_salesperson
```

## 1.4 Convert the Result into Clean DataFrame

```python
# Convert into a Normal DataFrame  Column.
region_sales_sorted = region_sales.sort_values(by='sales_amount', ascending=False)

region_sales_sorted
```

# Task 3: Custom Aggregation and Complete Sales Report

## 3.1 Create Custom Aggregation Function


```python
# Measure Sales variability with each region
def sales_range(x):
  return x.max()- x.min()
```

## 3.2 Apply Custom Funtion with Standard Aggregation

```python
# Generate detailed statistical summary  per region
region_analysis = df.groupby('region')['sales_amount'].agg(['sum','mean','min','max',sales_range]).reset_index()
region_analysis
```
